## Background

Sulfadoxine–pyrimethamine (SP) targets the parasite's folate pathway:

- **Sulfadoxine → Pfdhps**
- **Pyrimethamine → Pfdhfr**

Mutations in these genes reduce drug susceptibility and promote antifolate resistance.

## Aim

The study reviewed the **distribution and temporal trends of Pfdhfr and Pfdhps resistance markers** in *P. falciparum* across **East and West Africa from 2010–2025**.

## Methods

- Systematic review following **PRISMA 2020**
- **95 studies** included
- **9 East African** and **16 West African** countries
- Resistance-associated mutations and haplotypes were compared over time.

## Key Findings

### East Africa

High prevalence of:

**Pfdhfr:** N51I + C59R + S108N  
**Pfdhps:** A437G + K540E  

Wildtype: https://www.rcsb.org/3d-view/3QGT
mutatant complexed with pyrimethamin https://www.rcsb.org/3d-view/1J3J

These combinations contribute to the classical **quintuple resistance haplotype**. Additional **A581G** mutations were also detected.

### West Africa

The **Pfdhfr triple mutant** was also common, but **K540E was less prevalent**. West Africa showed greater diversity in Pfdhps mutations.

## Main Conclusion

Antifolate resistance is **widespread but geographically heterogeneous**. East Africa shows a greater accumulation of high-order **Pfdhfr/Pfdhps resistance haplotypes**, highlighting the need for continued molecular surveillance.

## Relevance to Drug Design

The study demonstrates why **pathogen mutations should be considered during computational drug discovery**. A useful workflow is:

**Wild-type target → resistant mutants → docking → MD simulation → MM/PBSA → compare binding stability**

This approach can determine whether a candidate drug or peptide retains activity against resistance-associated variants.

## What is the Folate Pathway?

The **folate pathway** is a biochemical pathway that produces and uses **folate derivatives**, which are essential for parasite growth and replication.

In *Plasmodium falciparum*, folate metabolism is particularly important for:

- **DNA synthesis**
- **RNA synthesis**
- **Amino-acid metabolism**
- **Purine synthesis**
- **Cell division**

The pathway ultimately provides **tetrahydrofolate (THF) derivatives**, which carry **one-carbon units** required for these essential cellular processes.

### Why is it important in malaria?

Folate metabolism is essential for the rapid growth and multiplication of *P. falciparum*. Therefore, enzymes involved in this pathway provide important **drug targets** for antimalarial therapy.

Two key enzymes are:

| Enzyme | Full name | Major drug target |
|---|---|---|
| **DHFR** | Dihydrofolate reductase | Pyrimethamine |
| **DHPS** | Dihydropteroate synthase | Sulfadoxine |

Together, these enzymes form the basis of the **folate pathway targeted by sulfadoxine–pyrimethamine (SP)**.
