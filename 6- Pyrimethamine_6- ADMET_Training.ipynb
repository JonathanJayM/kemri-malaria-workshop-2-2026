## 1. Introduction to ADMET in Drug Discovery

### What is ADMET?
In modern pharmacology and computer-aided drug design (CADD), **ADMET** stands for **Absorption, Distribution, Metabolism, Excretion, and Toxicity**. 

* **Absorption:** How the drug enters systemic circulation from its site of administration (e.g., intestinal mucosa following oral ingestion).
* **Distribution:** How the drug disperses throughout tissue spaces, blood plasma, biological fluids, and cellular compartments.
* **Metabolism:** How biotransformation enzymes (predominantly hepatic Cytochrome P450 enzymes) chemically alter the parent compound into metabolites.
* **Excretion:** How the molecule and its metabolites are eliminated from the organism (renal clearance, biliary/fecal excretion, pulmonary exhalation).
* **Toxicity:** The potential adverse organ reactions, mutagenicity, cellular damage, or off-target bioactivities produced by the chemical entity.

### Why is ADMET Profiling Critical?
Historically, over 50% of candidate drugs failed in clinical trials due to inadequate pharmacokinetics or unacceptable toxicity. Early computational (in silico) and cellular (in vitro) ADMET profiling allows medicinal chemists to identify and optimize promising compounds long before expensive animal trials or human clinical studies.

## 2. In-Depth Parameter Explanations

Below is a detailed breakdown explaining why every single parameter evaluated during molecular profiling is vital for transforming a small molecule into a safe, efficacious oral drug.

### A. Physicochemical Properties

#### 1. Molecular Weight (MW = 248.72 g/mol)
* **Why it matters:** MW directly influences molecular movement, diffusion rates through biological membranes, and steric occupancy within binding pockets. Lipinski's Rule of 5 sets $MW \le 500\text{ Da}$ for oral bioavailability.
* **Pyrimethamine Assessment:** At 248.72 Da, Pyrimethamine is a compact small molecule capable of rapid cellular diffusion and passive membrane transport.

#### 2. Molecular Volume (~215 Å³)
* **Why it matters:** Van der Waals volume reflects 3D spatial occupancy. Smaller volumes reduce steric hindrance during active transport or channel entry.
* **Pyrimethamine Assessment:** Fits precisely into the constrained active site cavity of protozoan Dihydrofolate Reductase (DHFR).

#### 3. Hydrogen Bond Acceptors (nHA = 4.0) & Donors (nHD = 2.0)
* **Why it matters:** Hydrogen bonds govern binding specificity with amino acid side chains. However, too many polar H-bond features require high energy to strip water molecules (desolvation penalty) before entering lipid bi-layers. Lipinski criteria: $nHA \le 10$, $nHD \le 5$.
* **Pyrimethamine Assessment:** Possesses 4 acceptor nitrogens (pyrimidine ring + amines) and 2 donor amino groups ($-\text{NH}_2$), optimal for target interaction without impeding membrane permeability.

#### 4. Rotatable Bonds (nRot = 2.0)
* **Why it matters:** Rotatable single bonds reflect molecular flexibility. High flexibility ($nRot > 10$) incurs a severe thermodynamic entropic penalty upon target binding and correlates with poor oral absorption.
* **Pyrimethamine Assessment:** Extremely rigid ($nRot = 2$), retaining a fixed 3D bio-conformation that enhances binding affinity.

#### 5. Topological Polar Surface Area (TPSA = 77.82 Å²)
* **Why it matters:** TPSA measures the surface area sum over polar atoms (N, O, and attached H). $TPSA < 140\text{ \u00c5}^2$ is required for intestinal absorption; $TPSA < 90\text{ \u00c5}^2$ is necessary to cross the Blood-Brain Barrier (BBB).
* **Pyrimethamine Assessment:** At 77.82 \u00c5\u00b2, Pyrimethamine possesses high intestinal absorption and effective BBB penetration.

#### 6. Lipophilicity (logP = 2.52)
* **Why it matters:** LogP is the octanol/water partition coefficient. Compounds with $\text{logP} < 0$ are overly hydrophilic (cannot cross lipid membranes); compounds with $\text{logP} > 5$ are overly lipophilic (poor aqueous solubility, metabolic instability, non-specific binding).
* **Pyrimethamine Assessment:** A logP of 2.52 sits directly in the golden window ($1.0 - 3.0$) for ideal oral drug leads.

#### 7. Ionization Constant (pKa = 7.03)
* **Why it matters:** Determines ionization state across pH gradients (stomach pH 1–2, intestine pH 6–7.4, blood pH 7.4). Neutral species diffuse through lipid membranes, while charged species increase aqueous solubility.
* **Pyrimethamine Assessment:** Weak base; existing predominantly as a neutral species at physiological pH, supporting gastrointestinal absorption.

### B. Medicinal Chemistry & Drug-Likeness Rules

* **Lipinski’s Rule of 5:** Accepted ($MW \le 500$, $\text{logP} \le 5$, $nHD \le 5$, $nHA \le 10$).
* **Pfizer Rule (3/75 Rule):** Accepted ($\text{logP} < 3$ and $TPSA > 75\text{ \u00c5}^2$), predicting reduced off-target toxicity and high safety margins.
* **GSK Rule:** Accepted ($MW \le 400$, $\text{logP} \le 4$).
* **PAINS (Pan-Assay Interference Compounds):** 0 Alerts. Free of reactive groups that cause false positives in high-throughput assay screening.

---

### C. ADME & Toxicity Profile

* **Absorption (Caco-2 / HIA):** Rapid and nearly complete oral absorption (~90–100% oral bioavailability).
* **Distribution (BBB / PPB):** Crosses the Blood-Brain Barrier (crucial for treating cerebral toxoplasmosis). Moderately bound to plasma proteins (~80–87%).
* **Metabolism:** Metabolized in the liver via Cytochrome P450 enzymes (primarily CYP3A4 and CYP2C9).
* **Excretion & Half-Life:** Displays a notably long elimination half-life (~80–100 hours in humans), enabling once-daily or weekly dosing regimens.
* **Toxicity Considerations:** Primary toxicity is dose-dependent folic acid deficiency due to host DHFR inhibition, clinically managed with leucovorin (folinic acid) supplementation.

```python
# ========================================================================= #
# 3. Executable Python Code: RDKit Descriptor Calculation & Visualization   #
# ========================================================================= #

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from rdkit import Chem
from rdkit.Chem import Draw, Descriptors

# 1. Define Pyrimethamine SMILES
smiles = "CCC1=C(C(=NC(=N1)N)N)C2=CC=C(C=C2)Cl"
mol = Chem.MolFromSmiles(smiles)

# 2. Render 2D Chemical Structure
img = Draw.MolToImage(mol, size=(500, 300))
plt.figure(figsize=(7, 4))
plt.imshow(img)
plt.axis('off')
plt.title("Pyrimethamine: 5-(4-chlorophenyl)-6-ethylpyrimidine-2,4-diamine", fontsize=11, fontweight='bold')
plt.show()

# 3. Compute RDKit Molecular Descriptors
computed_data = {
    "Parameter": [
        "Molecular Weight (g/mol)",
        "LogP (Lipophilicity)",
        "TPSA (\u00c5\u00b2)",
        "H-Bond Donors (nHD)",
        "H-Bond Acceptors (nHA)",
        "Rotatable Bonds (nRot)"
    ],
    "RDKit Calculated Value": [
        round(Descriptors.MolWt(mol), 2),
        round(Descriptors.MolLogP(mol), 2),
        round(Descriptors.TPSA(mol), 2),
        Descriptors.NumHDonors(mol),
        Descriptors.NumHAcceptors(mol),
        Descriptors.NumRotatableBonds(mol)
    ],
    "Lipinski Rule Threshold": [
        "<= 500.0",
        "<= 5.0",
        "<= 140.0",
        "<= 5.0",
        "<= 10.0",
        "<= 10.0"
    ]
}

df = pd.DataFrame(computed_data)
print("\n=================== Pyrimethamine Physicochemical Verification ===================")
print(df.to_string(index=False))

# 4. Generate Multi-Axis ADMET Radar Chart
labels = ['LogP (Lipophilicity)', 'MW / 100', 'TPSA / 10', 'RotBonds', 'H-Acceptors', 'H-Donors']
values = [
    Descriptors.MolLogP(mol),
    Descriptors.MolWt(mol) / 100,
    Descriptors.TPSA(mol) / 10,
    Descriptors.NumRotatableBonds(mol),
    Descriptors.NumHAcceptors(mol),
    Descriptors.NumHDonors(mol)
]

num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
values += values[:1]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
ax.plot(angles, values, color='#2ca02c', linewidth=2.5, linestyle='solid', label='Pyrimethamine Profile')
ax.fill(angles, values, color='#2ca02c', alpha=0.25)

ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels, fontsize=10, fontweight='bold')
ax.set_title("Pyrimethamine ADME & Drug-Likeness Radar Chart", size=13, weight='bold', y=1.1)
plt.legend(loc='upper right', bbox_to_anchor=(1.25, 1.1))
plt.show()
```
