In molecular docking, sampling algorithms and scoring functions are the two main components you should focus on. Sampling generates possible ligand-binding poses, while scoring evaluates and ranks those poses.

## Sampling Algorithms

#### Monte Carlo (MC) sampling 

Randomly changes ligand position, orientation or torsion angles. A new pose is accepted or rejected according to an acceptance rule, often the Metropolis criterion.

#### Genetic algorithm (GA)

Treats candidate poses as individuals in a population. Selection, crossover and mutation generate new candidates over successive generations

#### Lamarckian genetic algorithm (LGA)

Combines genetic search with local optimization. Locally improved poses can influence subsequent generations

## Scoring Algorithms

#### Empirical scoring functions

Combine weighted terms for interactions such as hydrogen bonding, hydrophobic contacts, steric complementarity and torsional penalties.

Examples: AutoDock Vina scoring functions, ChemScore, GlideScore.

```bash

#### Step 5: Energy Minimization

Perform geometry optimization on all generated .sdf ligand files using the MMFF94 force field for up to 1000 iterations:



```bash
mkdir -p minimized
for f in *.sdf; do
    [ -f "$f" ] || continue
    out="minimized/${f%.sdf}_minimized.sdf"
    obabel "$f" -O "$out" \
        --minimize --ff MMFF94 --steps 1000
done

#### Step 6: Convert Ligand Structures to PDBQT Format

Batch convert the minimized .sdf files into .pdbqt format (required for AutoDock Vina input):

```bash
for f in *.sdf; do
    obabel "$f" -O "${f%.sdf}.pdbqt"
done
